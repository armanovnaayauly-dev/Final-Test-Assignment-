```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Данные для тестового задания - Данные об аудитории.csv')
```

```python
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
start_date = '2023-11-01'
end_date = '2023-11-30'
mask = (df['date'] >= start_date) & (df['date'] <= end_date)
november_data = df[mask]
```

```python
mau_november = november_data['user_id'].nunique()
print("MAU за ноябрь 2023:", mau_november)
```

```python
dau = df.groupby('date')['user_id'].count().reset_index()
print(dau)
```

```python
cohort_date = '2023-11-01'
cohort_users = df[df['date'] == cohort_date]['user_id'].unique()

# Проверяем, сколько из них вернулись 2 ноября
next_day = pd.to_datetime(cohort_date) + pd.Timedelta(days=1)
returned_users = df[(df['date'] == next_day) & (df['user_id'].isin(cohort_users))]['user_id'].nunique()

# D1 Retention в %
d1_retention = (returned_users / len(cohort_users)) * 100
print(f"D1 Retention для пользователей, пришедших {cohort_date}: {d1_retention:.2f}%")
```

```python
total_users = november_data['user_id'].nunique()
converted_users = november_data[november_data['view_adverts'] > 0]['user_id'].nunique()
conversion = (converted_users / total_users) * 100
print(conversion)
```

```python
views_per_user = november_data.groupby('user_id')['view_adverts'].sum()
avg_views=views_per_user.mean()
print(avg_views)
```
